```python
import xmlcoll as xc
```

Create a collection.

```python
coll = xc.coll.Collection()
```

Add item and properties.

```python
item = xc.coll.Item("Enstatite")
properties = ({('mu', 'ti50', 'value'): 50,
               ('mu', 'ti50', 'error'): 7,
               ('mu', 'ti50', 'error type'): "2sd",
               ('mu', 'ca48', 'value'): 40,
               ('mu', 'ca48', 'error'): 10,
               ('mu', 'ti50', 'error type'): "2sd",
               ('note', '1'): "This is the first note.",
               ('note', '2'): "This is the second note.",
               "source": "Amazing paper (2020)"})
item.update_properties(properties)
coll.add_item(item)

item = xc.coll.Item("Mars")
properties = ({('mu', 'ti50', 'value'): 40,
               ('mu', 'ti50', 'error'): 6,
               ('mu', 'ti50', 'error type'): "2sd",
               ('mu', 'ca48', 'value'): 50,
               ('mu', 'ca48', 'error'): 10,
               ('mu', 'ti50', 'error type'): "2sd",
               "source": "Another Amazing paper (2021)"})
item.update_properties(properties)
coll.add_item(item)
```

Output collection to XML.

```python
output_file = "my_file.xml"
coll.write_to_xml(output_file)
```

Output to excel.

```python
df = coll.get_dataframe()
df.to_excel('my_file.xlsx')
```
