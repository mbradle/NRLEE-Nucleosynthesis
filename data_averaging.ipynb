# Define routines

Define a routine to compute the weighted mean and variation.

```python
def compute_mean_and_error(data, errors):
    assert len(data) == len(errors), "Data and errors arrays must be same size."
    for error in errors:
        assert error != 0, "No error can be zero."
        
    weights = 1 / np.array(errors)**2
    return (np.average(data, weights=weights), 1. / np.sqrt(np.sum(weights)))
```

# Calculation

In this section, compute the mean and variation for the set of data.  Begin by entering the data and errors.

```python
data = np.array([-4, -3, -4])
errors = np.array([15, 17, 17])
```

Now compute the mean and the variation and print the results.

```python
mean, variation = compute_mean_and_error(data, errors)

print(f"Weighted Mean: {mean:.4f}")
print(f"Variation: {variation:.4f}")
```
