# Define routines

Define a routine to compute the mean.

```python
def compute_mean(data, errors, mean_type):
    match mean_type:
        case "weighted":
            return sum(value / error**2 for value, error in zip(data, errors)) / sum(1. / error**2 for error in errors)
        case "straight":
            return np.mean(data)
        case _:
            print("Invalid mean type")
            return None
```

Define a routine to compute the confidence interval.  Input the data, errors, and confidence level.  The routine returns the lower and upper bounds on the confidence interval.

```python
def confidence_interval(data, errors, conf_level, mean_type):
    mean = compute_mean(data, errors, mean_type)
    sem = np.sqrt(np.sum(errors**2) / len(data))
    interval = st.t.interval(conf_level, len(data)-1, loc=mean, scale=sem)
    return interval
```

# Statistics choices

In this section, make your statistics choices.  First, choose the mean type ("straight" or "weighted").

```python
mean_type = "weighted"
```

Next choose the confidence level.

```python
conf_level = 0.95
```

# Calculation

In this section, compute the mean and variation for the set of data.  Begin by entering the data and errors.

```python
data = np.array([10, 12, 15, 14, 18])
errors = np.array([1, 1.5, 0.8, 1.2, 1])
```

Now compute the mean and the variation and print the results.

```python
mean = compute_mean(data, errors, mean_type)
result = confidence_interval(data, errors, conf_level, mean_type)

print(f"Weighted Mean: {mean:.4f}")
print(f"Variation: {mean - result[0]:.4f}")
```
