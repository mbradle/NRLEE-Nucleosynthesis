# Define routines

Define the routine to compute the confidence interval.  Input the data, errors, and confidence level.  The routine returns the mean and the lower and upper bounds on the confidence interval.

```python
def mean_and_confidence_interval(data, errors, conf_level):
    mean = np.mean(data)
    sem = st.sem(data) # Standard error of the mean
    if errors is not None:
      sem = np.sqrt(np.sum(errors**2) / len(data)) # Combine errors if provided
    
    interval = st.t.interval(conf_level, len(data)-1, loc=mean, scale=sem)
    return (mean, interval)
```

# Calculations

Enter data and errors.

```python
data = np.array([10, 12, 15, 14, 18])
errors = np.array([1, 1.5, 0.8, 1.2, 1])
```

Choose the confidence level.

```python
conf_level = 0.95
```

Compute confidence interval with errors.

```python
result = mean_and_confidence_interval(data, errors, conf_level)
print(f"Mean: {result[0]:.4f}")
print(f"95% Confidence Interval: {result[1][0]:.4f} -- {result[1][1]:.4f}")
print(f"Variation: {result[0] - result[1][0]:.4f}")
```

Compute confidence interval without errors.

```python
result = mean_and_confidence_interval(data, None, conf_level)
print(f"Mean: {result[0]:.4f}")
print(f"95% Confidence Interval: {result[1][0]:.4f} -- {result[1][1]:.4f}")
print(f"Variation: {result[0] - result[1][0]:.4f}")
```
