Next, import the necessary packages:

```python
import os, io, requests
import xmlcoll as xc
```

First retrieve the meteoritic data.

```python
my_collection = xc.coll.Collection()
my_collection.update_from_xml(io.BytesIO(requests.get('https://osf.io/wj5rd/download').content))
#my_collection.update_from_xml('leya_et_al_2009_Ti.xml')
```

Print out global properties.

```python
global_props = my_collection.get_properties()
species = []
for key, value in global_props.items():
    print(key, ':', value)
```

Print out all items in collection.

```python
for item in my_collection.get():
    print(item)
```

Print all properties for selected item.

```python
name = "Brachinite"

body = my_collection.get()[name]
props = body.get_properties()
for k, v in props.items():
    print(k, ':', v)
```
