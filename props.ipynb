# Retrieve XML file and print out data

First retrieve the data from the XML file.  By default, all samples are parsed in.  To select a subset of the data, edit *my_xpath*.  The commented line shows an example.  Similarly, use a different input file by commenting the *update_from_xml* command and uncomment and modify the commented version.

```python
my_collection = xc.coll.Collection()

my_xpath = ""
#my_xpath = "[.//property['name' = 'type'] = 'NC']"

my_collection.update_from_xml(io.BytesIO(requests.get('https://osf.io/wj5rd/download').content), xpath = my_xpath)
#my_collection.update_from_xml('leya_et_al_2009_Ti.xml')
```

Print out global properties for the file.

```python
global_props = my_collection.get_properties()
species = []
for key, value in global_props.items():
    print(key, ':', value)
```

Write out all samples and their properties to a text file.  The file will be available in the local directory (on Colab, click on the tab icon on the left panel).  Change the name of the text file by changing the file string for *prop_file* (by default, set to *my_props_file.txt*).  To print the  file contents to the netebook, change *print_out* from *False* to *True*.

```python
prop_file = 'my_props_file.txt'
print_out = False

f = open(prop_file, "w")

for item in my_collection.get().values():
    f.write(f"Sample: {item.get_name()}\n\n")
    for k, v in item.get_properties().items():
        f.write(f"{k}: {v}\n")
    f.write("\n\n")

f.close()

if print_out:
    with open(prop_file, 'r') as f:
        print(f.read())

f.close()
```
